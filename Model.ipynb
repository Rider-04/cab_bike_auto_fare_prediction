```python
import pandas as pd
```

```python
ola = pd.read_csv(r"C:\Users\lenovo\OneDrive\Desktop\Folder\cab_bike_auto_fare_prediction\Bengaluru_ola.csv")
bookings = pd.read_csv(r"C:\Users\lenovo\OneDrive\Desktop\Folder\cab_bike_auto_fare_prediction\Bookings.csv")
ncr = pd.read_csv(r"C:\Users\lenovo\OneDrive\Desktop\Folder\cab_bike_auto_fare_prediction\ncr_ride_bookings.csv")
rider = pd.read_csv(r"C:\Users\lenovo\OneDrive\Desktop\Folder\cab_bike_auto_fare_prediction\rides_data.csv")
```

```python
ola_clean = ola.rename(columns={
    "Booking ID": "booking_id",
    "Vehicle Type": "vehicle_type",
    "Pickup Location": "pickup_location",
    "Drop Location": "drop_location",
    "Booking Value": "fare",
    "Ride Distance": "distance_km",
    "Payment Method": "payment_method",
    "Driver Ratings": "driver_rating",
    "Customer Rating": "customer_rating"
})
```

```python
rider_clean = rider.rename(columns={
    "services": "vehicle_type",
    "source": "pickup_location",
    "destination": "drop_location",
    "ride_id": "booking_id",
    "distance": "distance_km",
    "total_fare": "fare",
    "payment_method": "payment_method"
})
```

```python
ola_clean["date"] = pd.to_datetime(ola_clean["Date"], errors="coerce")
```

```python
ola_clean["hour"] = pd.to_datetime(
    ola_clean["Time"], errors="coerce"
).dt.hour

ola_clean["day_of_week"] = ola_clean["date"].dt.dayofweek
```

```python
final_df = pd.concat(
    [ola_clean, bookings_clean, ncr_clean, rider_clean],
    ignore_index=True
)
```
